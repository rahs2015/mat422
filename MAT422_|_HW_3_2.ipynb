# 3.2.1 Limits and Continuity
A llimit is the means for which we can derive conclusions about a graph's behavior. As seen below for the example, the function is technically undefineded at 0 but that is not to say that there is a behavior of the graph that allows us to describe what y position the graph is pointing towards for 0 as seen to be .5. This is all to describes a removable discontinuity. The graph is shown to be approaching .5 from both sides as we get closer and closer as described in the step sizes decreasing; it is seen from their respective y values that it is getting closer and closer to .5.

```python
h = np.array([0.5, 0.1, 0.05, 0.01, 0.001])
values = (1 - np.cos(h)) / h**2

print("Step sizes:", h)
print("Function values:", values)
print("Expected limit:", 0.5)

x = np.linspace(-2, 2, 400)
y = np.where(x != 0, (1 - np.cos(x)) / np.where(x != 0, x**2, 1), np.nan)

plt.plot(x, y)
plt.axvline(0, color="gray", linewidth=0.8)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Behavior near a removable discontinuity")
plt.grid()
plt.show()
```

# 3.2.2: Derivatives
Derivatives are used to describe the rate of change of a graph. In the example below using a quadratic equation with its tangent to describe the derivative as it shows its equivalence to being a derivative. The function of x is shown to be x^2 + 3x with its derivative being 2x + 3 @ 2 (pulling it to be y = 7). The tangent does not curve with the curve as its slope is to only describe the change for a specific point. Individuals should notice that the tangent line shall intersect with the curve but brieflly as it to accurately describe the change of the curve.

```python
def f(x):
    return x**2 + 3*x

point = 2
steps = np.array([0.5, 0.1, 0.01, 0.001])

estimates = (f(point + steps) - f(point - steps)) / (2 * steps)
slope = 2 * point + 3
y_point = f(point)
exact = 2 * point + 3

print("Estimated derivatives:", estimates)
print("Exact derivative:", slope)
print("Absolute errors:", np.abs(estimates - exact))

x = np.linspace(0, 4, 200)

plt.plot(x, f(x), label="f(x) = x^2 + 3x")
plt.plot(x, slope * (x - point) + y_point,
         "--", label="Tangent line at x = 2")
plt.scatter(point, y_point, label="Point (2, 10)")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Tangent Line and Derivative")
plt.legend()
plt.grid()
plt.show()
```

# 3.2.3: Taylor's Theorem
A taylor polynomial is used to approximates a function using its derivative. In this example, for the function of (x+1) ^1/2, the derivative of the function is used to follow the original function and account for its changes as a means to approximate the function. As the third degrees taylor gets farther away from 0 which is where we center our prediction we lose our ability to accuartely estiamte the function.

```python
x = np.linspace(-0.9, 2, 300)

actual = np.sqrt(1 + x)
approx = 1 + x / 2 - x**2 / 8 + x**3 / 16

plt.plot(x, actual, label="Actual sqrt(1 + x)")
plt.plot(x, approx, "--", label="Taylor approximation")
plt.xlabel("x")
plt.ylabel("Value")
plt.title("Taylor Approximation of a Square Root")
plt.legend()
plt.grid()
plt.show()

point = 0.5

actual_value = np.sqrt(1 + point)
approx_value = 1 + point / 2 - point**2 / 8 + point**3 / 16

print("Actual value:", actual_value)
print("Approximation:", approx_value)
print("Absolute error:", abs(actual_value - approx_value))
```
