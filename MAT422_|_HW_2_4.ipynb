#MAT 422 | HW 2.4: MLE and Linear Regresssion

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(422)
```

#2.4.1 MLE for Random Samples
Maximum Liklihood estimation (MLE) is used in order to find the parameter value that makes an observed sample most probable. The parameter (p) represents the probability of getting a success. The MLE is then found through dividing the intended object/success by the sample size. In the example below the highest point of the log-likelihood occurred around the calculated value of p. This shows that the sample had outputed the maximum likelihood estimate.

```python
tp = .6
n = 60
sample = rng.binomial( 1, tp, size = n)

suc = sample.sum()
phat = suc / n

grid = np.linspace(.01, .09, 200)

loglikehood = (
    suc * np.log(grid) + (n - suc) * np.log(1-grid)
)

print("Sample Size  = ", n)
print("Successes    = ", suc)
print("MLE for P    = ", phat)

plt.plot(grid, loglikehood)
plt.axvline(phat, linestyle = "--", label = "MLE")
plt.xlabel("p")
plt.ylabel("Log Likelihood")
plt.title("MLE for a Bernoulli Random Sample")
plt.legend()
plt.show()
```

#2.4.2 Linear Regression
Linear Regression is a means to describe a relationship between variables by which we estimate a line that minimizes the difference between the transcribed line and all the plots. This minimization is referred to the least squares/ squarred errror. Below a set of data is generated with positive correlation between 2 variables with small variation. The regression line is show to trace over some data points as well as residing decently in between all of them which fits the described notion of what a regression line is meant to be.

```python
x = np.linspace(0,10,20)
y = 3 + 1.8 * x + rng.normal(0,2,size = x.size)

X = np.column_stack([np.ones_like(x), x])
bhat = np.linalg.inv(X.T @ X) @ X.T @ y

yhat = X @ bhat
resid = y - yhat

print("Est. intercept and slope = ", bhat)
print("Resid sum of squares = ", np.sum(resid ** 2))

plt.scatter(x,y,label = "Data")
plt.plot(x,yhat, label = "Regression line")
plt.xlabel ("x")
plt.ylabel("y")
plt.title("Linear Regression using least squares")
plt.legend()
plt.show()
```
