# 1.2.3 Gram-Schmidt process
The Gram-Schmidt process uses a set of independent vectors and creates an orthonormal basis while keeping the span of the original vectors. A 3x3 matrix was generated and ran through the Gram-Schmidt process to output its Q matrix. At the end, the output was backed by its q.t*q aproximating the idenity matrix and dot products = 0; its norm checks ≈ 1 checking the length, and the reconstructions matches well with the original matrix.

```python
a = [1.0, 1.0, 0.0]
b = [1.0, 1.0, 1.0]
c = [0.0, 1.0, 1.0]

def gram(i,i1,i2):

  v = np.array(i, dtype = float)
  q = v / np.linalg.norm(v)
  print("v1 =", v)

  x1 = np.array(i1, dtype = float)
  v1 = x1 - (np.dot(x1,v)/np.dot(v,v))*v
  q1 = v1 /np.linalg.norm(v1)
  print("v2 =", v1)

  x2 = np.array(i2, dtype = float)
  v2 = x2 - (np.dot(x2,v)/np.dot(v,v))*v - (np.dot(x2,v1)/np.dot(v1,v1))*v1
  q2 = v2 /np.linalg.norm(v2)
  print("v3 =", v2)

  return np.column_stack((q,q1,q2))

Q = gram(a,b,c)



print ("\n\nThe orthonormal basis: \n", Q)
print ("\n\nQ.T @ Q:")
print(Q.T @ Q)

print("\n\nClose to the Identity matrix: ")
print("   ",np.allclose(Q.T @ Q, np.eye(3)))

print("\nNormCheck:")
for i in range(Q.shape[1]):
    print(f"    Norm of q{i+1}: {np.linalg.norm(Q[:,i])}")

print("\nOrthogonality Check:")
print("   ", np.dot(Q[:,0], Q[:,1]))
print("   ", np.dot(Q[:,0], Q[:,2]))
print("   ", np.dot(Q[:,1], Q[:,2]))

print("\nSpan Check")
X = np.column_stack((a,b,c))
R = Q.T @ X
X_rec= Q @ R
print("   ",np.allclose(X,X_rec))
```

#1.2.4 Eigenvalues and Eigenvectors
An eigenvector is a ≠0 vector that when a matrix is applied its direction is unchanged. Though since the direction is unchanged, the scale of the vector must be different. This bringsforth the equation of Av = λv. λ is the eigenvalue which is meant to describe the change of the vector's scale while v is the eigenvector and A is the matrix. The eigenvectors in the section below come out to be 3 and 2 with eigenvectors describing their direction of x and y respectively.

Trace and determinants were added as a differentiator as well as a check as the sum of eigenvalues equals the matrix's trace and the determinant is the product of the eigenvalues.

```python
a1 = np.array([
    [3,0],
    [0,2]
], dtype = float)


eigvl,eigvc = np.linalg.eig(a1)
print("Eigenvalues: ", eigvl)
print("   These values are what scale the vectors")
print("\nEigenvectors:\n", eigvc)

av = a1@eigvc
print(f"\nA @ v:\n{av}")

lv = eigvl * eigvc
print(f"\nλ * v:\n{lv}")

resid = av-lv
residNorm = np.linalg.norm(resid)
print(f"Residual Norm: {residNorm}")
if np.isclose(residNorm,0):
  print("   Statement 'Av = λ * v' holds true")
else:
  print("Highkey no clue")


for j in range(len(eigvl)):

  print(f"\nFor Eigenvalue:  {eigvl[j]}")
  print(f"   Eigenvector: {eigvc[:,j]}")
  print(f"   A v:         {av[:,j]}")
  print(f"   λ * v:       {lv[:,j]}")



maTrace= np.trace(a1)
print("\n\nTrace and Determinant\n")
print("  Trace of matrix:      ", maTrace)
print("  Sum of Eigenvalues:   ", np.sum(eigvl))

maDet = np.linalg.det(a1)
print("\n  Determinat of matrix: ", maDet)
print("  Product of Eigenvalues:   ", np.prod(eigvl))



eigdot = np.dot(eigvc[:,0], eigvc[:,1])
print("\n\nOrthogonality check:")
print("  Eigenvector dot product:", eigdot)

if np.isclose(eigdot,0):
  print("    - Eigenvectors are orthogonal")

```
