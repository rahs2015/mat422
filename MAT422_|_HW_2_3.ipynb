# MAT 422 | HW 2.3: Joint Distributions, Correlation, and Random Sampling

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(422)
```

#2.3.1 | Joint Probability Distrubtions
A joint probability distribtion describes the probability of 2 random variables occuring in tandem. The rows are described as X and as for the columns they are represented by Y. Adding across rows gives the marginal probabilities of X, while adding down gives the marginal probabilities of Y. These joint probabilites should add to 1 as to confirm the table being a properly represented probability distribution. The conditional probability P(y=2 | x=1) shows the probability of Y=2 when X = 1 has already occurred.

```python
joint = np.array([[0.12, 0.18, 0.10], [0.08, 0.22, 0.30]])

px = joint.sum(axis = 1)
py = joint.sum(axis = 0)

print("Joint total = ", joint.sum())
print("P(X) =", px)
print("P(Y) =", py)
print("P(Y=2 | X=1) =", joint[1, 2] / px[1])
```

#2.3.2 | Correlation and Dependence
Correlations is used to measure a graph's strength and direction between variables as shown below with 2 variables displaying what idealy a dependent and independent graph should typically look like. The values of correlation are between 1 to -1 with both polars sides representing strong correlation between each other with the negative displaying relationship. Dependence is still regardless not definitive for correlation as dependence can be shown without a storng linear relationship.  



```python
x = rng.normal (0,1,200)
yDe = 1.5 * x + rng.normal(0, 0.6, 200)
yIn = rng.normal(0,1, 200)

corrDep = np.corrcoef(x, yDe)[0,1]
corrIn = np.corrcoef(x, yIn)[0, 1]

print("Correlation with dependent variable =", corrDep)
print("Correlation with independent variable =", corrIn)

plt.scatter(x, yDe, alpha=0.7, label="dependent")
plt.scatter(x, yIn, alpha=0.5, label="independent")

plt.legend()
plt.title("Comparing Correlation and Dependence")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

#2.3.3 | Random Samples
A random sample consists of random observerations selected from a population. Using such a sampling process allows for minimal bias during sample selection which allows one to get a proper a sample of the population; however due to its inherit nature of being a sample repititon of such a method may display variation from previous samples done. The mean in this example should be close to the population's mean (3).



```python
sample_means = []

for _ in range(1000):
    sample = rng.exponential(scale=3.0, size=25)
    sample_means.append(sample.mean())

sample_means = np.array(sample_means)

print("Mean of sample means =", sample_means.mean())
print("Std of sample means =", sample_means.std())

plt.hist(sample_means, bins=30, edgecolor="white")
plt.title("Sampling Distribution of the Sample Mean")
plt.xlabel("Sample Mean")
plt.ylabel("Count")
plt.show()
```
